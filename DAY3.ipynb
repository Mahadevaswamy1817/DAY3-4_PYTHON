```python
weight = float(input("enter weight in kgs: "))
height_cm = float(input("enter height in cm: "))

# Convert height from cm to meters and calculate BMI
height_m = height_cm / 100
bmi = weight / (height_m ** 2)

if bmi < 18.5:
    print(f"Your BMI is {bmi:.2f}. You are underweight.")
else:
    print(f"Your BMI is {bmi:.2f}.")
```

```python
#leap year
year = int(input("enter the year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("leap year")
        else:
            print("not leap year")
    else:
        print("leap year")
else:
    print("not leap year")
```

```python
shoping_amount=int(input("enter amount: "))
if shoping_amount>=5000:
  discount_percent=20
  print("20 % discount")
elif shoping_amount>=3000:
  discount_percent=15
  print("15% discount")
elif shoping_amount>=100:
  discount_percent=10
  print("10% discount")
else:
  discount_percent=0
  print("no discount")

discount=shoping_amount*(discount_percent/100)
total_amount=shoping_amount-discount
print(f"total amount is {total_amount}")
```
